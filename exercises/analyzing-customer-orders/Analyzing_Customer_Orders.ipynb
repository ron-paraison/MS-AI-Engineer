```python

# This Cell Contains Data Structures that are Constructed for Analysis

# Customer Names in a List
customer_names = ["Abe", "Bob", "Cara", "Dolly", "Eve", "Fred", "Gina", "Hank", "Ivory", "Jerry"]

# List of Tuples Containing Order Details: (Customer Name, Product Name, Price, Category)
order_details = [
    ("Abe", "TV", 300, "Electronics"),
    ("Abe", "Tablet", 200, "Electronics"),
    ("Abe", "Blender", 90, "Home Appliances"),
    ("Bob", "Tablet", 200, "Electronics"),
    ("Cara", "Shirt", 20, "Clothing"),
    ("Cara", "Toaster", 60, "Home Appliances"),
    ("Cara", "Xbox Series X", 649.99, "Electronics"),
    ("Dolly", "Blender", 80, "Home Appliances"),
    ("Eve", "Shirt", 20, "Clothing"),
    ("Eve", "Hat", 10, "Clothing"),
    ("Fred", "Jeans", 30, "Clothing"),
    ("Fred", "Hat", 10, "Clothing"),
    ("Gina", "Toaster", 60, "Home Appliances"),
    ("Hank", "Tablet", 200, "Electronics"),
    ("Ivory", "Tablet", 200, "Electronics"),
    ("Ivory", "Toaster", 60, "Home Appliances"),
    ("Jerry", "Blender", 90, "Home Appliances"),
    ("Jerry", "Tablet", 200, "Electronics"),
    ("Jerry", "Jeans", 40, "Clothing")
]

# Dictionary with Customer Names as Keys and their Orders as Values
dictionary_of_orders = {
    "Abe": [ ("Abe", "TV", 300, "Electronics"), ("Abe", "Tablet", 200, "Electronics"), ("Abe", "Blender", 90, "Home Appliances")],
    "Bob": [("Bob", "Tablet", 200, "Electronics")],
    "Cara": [("Cara", "Shirt", 20, "Clothing"), ("Cara", "Toaster", 60, "Home Appliances"), ("Cara", "Xbox Series X", 649.99, "Electronics")],
    "Dolly": [("Dolly", "Blender", 80, "Home Appliances")],
    "Eve": [("Eve", "Shirt", 20, "Clothing"), ("Eve", "Hat", 10, "Clothing")],
    "Fred": [("Fred", "Jeans", 30, "Clothing"),("Fred", "Hat", 10, "Clothing")],
    "Gina": [("Gina", "Toaster", 60, "Home Appliances")],
    "Hank": [("Hank", "Tablet", 200, "Electronics")],
    "Ivory": [("Ivory", "Tablet", 200, "Electronics"), ("Ivory", "Toaster", 60, "Home Appliances")],
    "Jerry": [("Jerry", "Blender", 90, "Home Appliances"), ("Jerry", "Tablet", 200, "Electronics"), ("Jerry", "Jeans", 40, "Clothing")]
}
```

```python
# Classify Products by Category
products_by_category = {"Clothing": ["Hat", "Jeans", "Shirt", "Suit"],
                         "Electronics": ["Tablet", "TV", "Xbox Series X"],
                            "Home Appliances": ["Blender", "Microwave", "Toaster"]}

# Create a Set of Unique Product Categories
unique_categories = set(products_by_category.keys())

# Display All Available Product Categories
print(f"The available product categories are {unique_categories}.")
```

```python
# Use a Loop to Calculate the total amount spent by each customer
for customer, orderdetails in dictionary_of_orders.items():
    orderdetails = orderdetails if isinstance(orderdetails, list) else [orderdetails]
    customer_total_spend = sum(orders[2] for orders in orderdetails)

    # Identify and Display Customer Value Based on Total Spend
    if customer_total_spend < 50:
        print(f"{customer} spent {customer_total_spend} in total and is classified as a low-value buyer.")
    elif 50 <= customer_total_spend < 100:
        print(f"{customer} spent {customer_total_spend} in total and is classified as a moderate buyer.")
    else:
        print(f"{customer} spent {customer_total_spend} in total and is classified as a high-value buyer.")
```

```python
# Create an Empty Container (Dictionary) before a Loop to Store Aggregated Results of Calculations (Iterations)
customer_spending = {}

# Calculate the Total Revenue per Product Category and Store it in a Dictionary
for order in order_details:
    category = order[3]
    revenue_per_category = order[2]
    if category in customer_spending:
        customer_spending[category] += revenue_per_category
    else:
        customer_spending[category] = revenue_per_category

# Display Total Revenue per Product Category
for category in customer_spending:
    print(f"Total revenue for {category} is {customer_spending[category]}")
```

```python
# Extract Unique Products from All Orders Using a Set
unique_products = set()
for customer, orders in dictionary_of_orders.items():
    for order in orders:
        product_name = order[1]
        unique_products.add(product_name)

# Display Unique Products
print("Unique products from all orders are:", unique_products)
```

```python
# Identify Customers Who Purchased Electronics
for purchases in order_details:
    if purchases[3] == "Electronics":
        print(f"{purchases[0]} purchased an electronic item: {purchases[1]}")
```

```python
# Initialize a Dictionary and Create Loop to Calculate the Total Amount Spent by Each Customer
customer_totals = {}
for customer, orders in dictionary_of_orders.items():
    total_spent = sum(order[2] for order in orders)
    customer_totals[customer] = total_spent

# Create a Sorted List of Tuples (customer, total_spent) in Descending Order
list_of_top3_customers_and_spend = sorted(customer_totals.items(), key=lambda x: x[1], reverse=True)[:3]

# Identify and Display the Top 3 Customers
list_of_top3 = []
for top3 in list_of_top3_customers_and_spend:
    top3_customers = top3[0]
    list_of_top3.append(top3_customers)
print(f"The Top 3 Customers are: {list_of_top3}")
```

```python
# Print a Summary of Each Customer's Total Spending and Display their Classification
for customer, total_spent in customer_totals.items():
    if total_spent < 50:
        classification = "low-value buyer"
    elif 50 <= total_spent < 100:
        classification = "moderate buyer"
    else:
        classification = "high-value buyer"
    print(f"{customer} spent a total of {total_spent} and is classified as a {classification}.")
```

```python
# Use Set Oerations to Identify Customers Who Purchased from Multiple Categories
category_customers = {}
for customer, orders in dictionary_of_orders.items():
    categories = set(order[3] for order in orders)
    category_customers[customer] = categories
    if len(categories) > 1:
            print(f"{customer} purchased from multiple categories: {categories}")

# The Loop is re-created to specifically identify customers who purchased both Electronics and Clothing
for customer, orders in dictionary_of_orders.items():
    categories = set(order[3] for order in orders)
    category_customers[customer] = categories
    if "Electronics" in categories and "Clothing" in categories:
            print(f"{customer} purchased both Electronics and Clothing.")
```

```python
# Create a Dictionary to Store the Results of the Calculations Below
product_frequency = {}

# Calculate the Purchase Frequency of Each Product
for order in order_details:
    product_name = order[1]
    if product_name in product_frequency:
        product_frequency[product_name] += 1
    else:
        product_frequency[product_name] = 1

# Identify and Display the Top 3 Purchased Products
for product, frequency in product_frequency.items():
    top3_products = sorted(product_frequency.items(), key=lambda x: x[1], reverse=True)[:3]
print(f"The Top 3 Purchased Products are: {top3_products}")
```
